```python
import os
os.environ['USE_HF'] = '1'
os.environ['CUDA_VISIBLE_DEVICES'] = '0'

from swift.infer_engine import TransformersEngine, RequestConfig, InferRequest

model = 'qwen/Qwen3.5-2B'
engine = TransformersEngine(model)

prompt = (
    "Hello qwen, I wanted to ask you some questions. First one, what colours is the sky on foggy day?"
    "Second, do birds fly backwards? And the last one, What is the power of rocket launcher?"
)

temperatures = [0.0, 1.0, 10.0, 1000.0]

for temp in temperatures:
    print("=" * 60)
    print(f"TEST DLA TEMPERATURY: {temp}")
    print("=" * 60)

    request_config = RequestConfig(max_tokens=1024, temperature=temp)

    infer_requests = [
        InferRequest(messages=[{'role': 'user', 'content': prompt}])
    ]

    resp_list = engine.infer(infer_requests, request_config)

    print(resp_list[0].choices[0].message.content)
    print("\n")
```

Dla temperatury 0 i 1 odpowiedź była dokładna, a sama składnia była, można powiedzieć, naturalna. <br>
Dla temperatury 10 odpowiedź może składniowo miała sens do pewnego momentu, ale nie odnosiła się zupełnie do żadanych pytań. <br>
Natomiast dla temperatury 1000 odpowiedź była długim zlepkiem losowych słów nie tylko z języka angielskiego.
